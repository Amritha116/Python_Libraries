```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)        
plt.xlabel("X-axis")  
plt.ylabel("Y-axis")  
plt.title("Simple Line Plot")
plt.show()  
```

```python

data = [12, 15, 20, 20, 22, 23, 23, 23, 25, 30, 30, 32, 35, 40, 45, 50]

plt.hist(data, bins=8, color='skyblue', edgecolor='black')

plt.xlabel("Value Range")
plt.ylabel("Frequency")
plt.title("Histogram Example")
plt.show()
```

```python
data = [7, 8, 5, 6, 9, 12, 15, 14, 10, 8, 6, 7, 9, 11, 13, 12, 8, 7]

plt.boxplot(data, patch_artist=True, boxprops=dict(facecolor="lightblue"))
plt.xlabel("Dataset")
plt.ylabel("Values")
plt.title("Box Plot Example")
plt.show()
```

```python
data1 = [7, 8, 5, 6, 9, 12, 15, 14, 10]
data2 = [15, 18, 20, 22, 19, 25, 30, 28, 24]

plt.boxplot([data1, data2], patch_artist=True,
            boxprops=dict(facecolor="lightgreen"))
plt.xticks([1, 2], ["Group 1", "Group 2"])
plt.title("Comparison of Two Groups")
plt.show()
```

```python
import pandas as pd
data = ["apple", "banana", "apple", "orange", "banana", "apple"]

# Count occurrences
count = pd.Series(data).value_counts()

# Plot
plt.bar(count.index, count, color=["red", "yellow", "orange"])
plt.show()
```

```python
data={
    "Salary":[500000,45000,60000,900000,350000,750000]
}
df = pd.DataFrame(data)
```

```python
from matplotlib.lines import lineStyles
plt.plot(df["Salary"],color="blue",marker="*",linestyle=":",linewidth="2")
plt.grid()
plt.show
```

```python
cibil_scores = [650, 700, 720, 750, 780, 800]
salaries = [30000, 40000, 45000, 50000, 60000, 70000]

plt.scatter(cibil_scores, salaries, color='blue', marker='o')
plt.xlabel("CIBIL Score")
plt.ylabel("Salary")
plt.title("Salary vs CIBIL Score")
plt.grid(True)
plt.show()
```

```python
labels = ['Rent', 'Food', 'Travel', 'Savings']
sizes = [40, 25, 20, 15]  
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
explode = (0.1, 0, 0, 0)   
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%',
        startangle=90, explode=explode, shadow=True)

plt.title('Expense Breakdown')
plt.axis('equal')  
plt.show()
```

```python
import seaborn as sns
```

```python

tips = sns.load_dataset("tips")

sns.scatterplot(x="total_bill", y="tip", data=tips, hue="sex", style="time")

plt.title("Scatter plot of Total Bill vs Tip")
plt.show()
```

```python
tips = sns.load_dataset("tips")
sns.lineplot(x="total_bill", y="tip", data=tips)
plt.show()
```

```python
sns.barplot(x="total_bill", y="tip", data=tips)
plt.show()
```

```python
sns.distplot(tips["total_bill"], kde=True)
plt.show()
```
